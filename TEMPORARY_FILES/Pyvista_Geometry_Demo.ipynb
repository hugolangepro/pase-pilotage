In this example an array of similar photovoltaic pannel using pyvista. The initial model is multiply over space using regular grids. 

Importation of the required module

```python
import numpy as np
import pyvista as pv
```

Creation of the initial solar panel which will be copy to constitute the pannel array.
Creation of the ground (only for visualization purpose).
The geometry has to be triangle in order to be able to cast multiple rays.

```python
#Solar pannel (flat rectangle)
vertices = np.array([[0, 0, 0], [1, 0, 0], [0, 1, 0], [1, 1, 0]])
faces = np.hstack(
    [
        [3, 0, 1, 2],  # triangle
        [3, 1, 2, 3],  # triangle
    ]
    )
sPV = pv.PolyData(vertices, faces)
sPV.rotate_x(35)


#Ground
vGround = np.array([[-10, -10, -1], [10, -10, -1], [-10, 10, -1], [10, 10, -1]])
fGround = np.hstack(
    [
        [4, 0, 1, 3,2],  # Rectangle
    ]
)
sGround = pv.PolyData(vGround,fGround)


plotter = pv.Plotter(window_size=(600, 400))
plotter.background_color = 'w'
plotter.enable_anti_aliasing()
plotter.add_mesh(sPV,color='black')
plotter.add_mesh(sGround,color='green')
plotter.show(jupyter_backend='pythreejs')
```

Creation of a PolyData array using the glyph method of a StructuredGrid (can also be done for UnstructuredGrid [cf](https://docs.pyvista.org/api/core/_autosummary/pyvista.UnstructuredGrid.glyph.html#pyvista.UnstructuredGrid.glyph)

```python
# Grid creation using numpy.mgrid 
x,y,z = np.mgrid[-3:3, -3:3, 0:1]
Localmesh = pv.StructuredGrid(x, y, z)

# Creation of the PV array, in this example the initial model is scale down by 25% 
LocalPVArray = Localmesh.glyph(geom=sPV, factor=0.75,scale = True)

plotter = pv.Plotter(window_size=(600, 400))
plotter.background_color = 'w'
plotter.enable_anti_aliasing()
plotter.add_mesh(LocalPVArray,color='black')
plotter.add_mesh(sGround,color='green')
plotter.show(jupyter_backend='pythreejs')
```

Creation of an array of array ! Using a larger structured grid

```python
#Creation d'une grille globale sur laquelle on va cloner l'ensemble local
x,y,z = np.mgrid[-3:3, -3:3, 1:2]*5
GlobalMesh = pv.StructuredGrid(x, y, z)
GlobalPVArray  = GlobalMesh.glyph(geom=LocalPVArray, factor=0.75,scale = True)

plotter = pv.Plotter(window_size=(600, 400))
plotter.background_color = 'w'
plotter.enable_anti_aliasing()
plotter.add_mesh(GlobalPVArray,color='black')
plotter.add_mesh(sGround,color='green')
plotter.show(jupyter_backend='pythreejs')
```
